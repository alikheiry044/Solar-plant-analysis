۱. مقایسه دمای محیط و دمای ماژول در طول روز

نمودار اول تغییرات دمای محیط و دمای ماژول را در ساعات مختلف شبانه‌روز برای نیروگاه اول نشان می‌دهد.

بر اساس نتایج مشاهده می‌شود که با افزایش تابش خورشیدی در طول روز، هر دو پارامتر دمای محیط و دمای ماژول افزایش می‌یابند. با این حال، نرخ افزایش دمای ماژول به‌مراتب بیشتر از دمای محیط است. دمای محیط در بیشترین مقدار خود به حدود ۳۰ درجه سانتی‌گراد می‌رسد، در حالی که دمای ماژول از ۵۰ درجه سانتی‌گراد نیز فراتر می‌رود.

این اختلاف نشان می‌دهد که پنل‌های خورشیدی در اثر جذب انرژی خورشید، گرمای قابل توجهی تولید می‌کنند و دمای آن‌ها بسیار بیشتر از دمای محیط اطراف می‌شود.

مقایسه تولید انرژی

میانگین تولید روزانه انرژی در نیروگاه اول برابر با 622.65 مگاوات‌ساعت در روز و در نیروگاه دوم برابر با 480.41 مگاوات‌ساعت در روز بوده است.

این نتایج نشان می‌دهد که تولید انرژی نیروگاه دوم تنها حدود 77 درصد تولید نیروگاه اول است. به عبارت دیگر، نیروگاه اول در طول دوره مورد بررسی عملکرد بهتری از نظر تولید انرژی داشته و توانسته است برق بیشتری تولید کند.

مقایسه شرایط آب‌وهوایی

بررسی داده‌های محیطی نشان می‌دهد که میانگین دمای نیروگاه اول 25.5 درجه سانتی‌گراد و میانگین دمای نیروگاه دوم 28.1 درجه سانتی‌گراد بوده است. بنابراین نیروگاه دوم به طور متوسط در شرایط دمایی گرم‌تری فعالیت کرده است.

از سوی دیگر، میانگین تابش خورشیدی ثبت‌شده در نیروگاه اول 0.2283 و در نیروگاه دوم 0.2327 بوده است. این مقادیر بسیار نزدیک به یکدیگر هستند و نشان می‌دهند که هر دو نیروگاه تقریباً میزان مشابهی از انرژی خورشیدی را دریافت کرده‌اند.

نتیجه‌گیری نهایی

با وجود آن‌که میزان تابش خورشیدی در دو نیروگاه تقریباً یکسان بوده است، نیروگاه اول انرژی بیشتری تولید کرده است. یکی از مهم‌ترین دلایل این اختلاف می‌تواند دمای پایین‌تر ماژول‌ها و شرایط حرارتی مناسب‌تر در نیروگاه اول باشد. افزایش دمای پنل‌های خورشیدی معمولاً موجب کاهش راندمان تبدیل انرژی می‌شود و این موضوع می‌تواند علت بخشی از کاهش تولید در نیروگاه دوم باشد.

در مجموع، نتایج این مطالعه نشان می‌دهد که نیروگاه اول از نظر بهره‌وری و عملکرد کلی وضعیت مطلوب‌تری نسبت به نیروگاه دوم داشته است و کنترل دمای عملیاتی پنل‌ها نقش مهمی در بهبود راندمان سیستم‌های فتوولتائیک ایفا می‌کند.
